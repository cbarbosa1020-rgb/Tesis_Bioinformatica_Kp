```python
import os
import pandas as pd
import requests
import time

# 1. Definir las rutas dentro de tu Google Drive montado
ruta_csv = "/content/drive/MyDrive/Proyecto_Tesis/metadatos/matriz_completa_cmi.csv"
carpeta_salida = "/content/drive/MyDrive/Proyecto_Tesis/secuencias"

print("-> Validando rutas en Google Drive...")

# Verificar que la matriz realmente exista en el Drive
if not os.path.exists(ruta_csv):
    print(f"[ERROR]: No se encontró la matriz en la ruta: {ruta_csv}")
    print("Por favor, verifica que la carpeta se llame 'Proyecto_Tesis' y el archivo esté dentro de 'metadatos'.")
else:
    # 2. Cargar los metadatos clínicos y aislar los IDs únicos de los genomas
    df = pd.read_csv(ruta_csv)
    genomas = df['genome_id'].unique()
    total_genomas = len(genomas)
    print(f"-> ¡Éxito! Matriz cargada correctamente.")
    print(f"-> Se procesarán un total de {total_genomas} genomas únicos de Klebsiella pneumoniae.\n")
    print("-> Iniciando descarga masiva directa a tu Drive de 1 TB (este proceso tomará tiempo)...")
    print("="*70)

    # 3. Bucle de descarga interactivo
    for i, genoma_id in enumerate(genomas):
        archivo_salida = os.path.join(carpeta_salida, f"{genoma_id}.fasta")

        # Sistema de reanudación segura: si el archivo ya existe en tu Drive, se lo salta
        if os.path.exists(archivo_salida):
            continue

        # URL oficial de la API de BV-BRC para extraer el formato FASTA de nucleótidos
        url = f"https://www.bv-brc.org/api/genome_sequence/?eq(genome_id,{genoma_id})&limit(25000)"
        headers = {"Accept": "application/dna+fasta"}

        try:
            respuesta = requests.get(url, headers=headers)
            if respuesta.status_code == 200:
                # Guardar el contenido directamente en la carpeta de secuencias de Drive
                with open(archivo_salida, "w") as archivo:
                    archivo.write(respuesta.text)
                print(f"[{i+1}/{total_genomas}] Descargado y guardado seguro: {genoma_id}.fasta")
            else:
                print(f"[{i+1}/{total_genomas}] Error en el servidor de BV-BRC para el ID {genoma_id} (Status: {respuesta.status_code})")
        except Exception as e:
            print(f"[{i+1}/{total_genomas}] Error crítico de conexión en ID {genoma_id}: {str(e)}")

        # Pausa de seguridad de 0.2 segundos para mantener estable la API de BV-BRC
        time.sleep(0.2)

    print("\n" + "="*70)
    print("-> ¡PROCESO FINALIZADO EN LA NUBE!")
    print(f"-> Todos los archivos .fasta disponibles se encuentran a salvo en: {carpeta_salida}")
```
